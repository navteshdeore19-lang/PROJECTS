```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.cluster import KMeans
```

```python
df = pd.read_csv('Mall_Customers.csv')
df
```

```python
X = df.iloc[:,[3,4]]
```

```python
X.head()
```

```python
df.shape
```

```python
# Find Best K (Elbow Method)

wcss = []

for i in range(1,11):

    model = KMeans(
        n_clusters=i,
        random_state=42
    )

    model.fit(X)

    wcss.append(model.inertia_)
```

```python
# Plot Elbow Graph

plt.plot(range(1,11), wcss)

plt.xlabel("Clusters")

plt.ylabel("WCSS")

plt.title("Elbow Method")

plt.show()

```

```python
# Train K-Means

model = KMeans(
    n_clusters=5,
    random_state=42
)

y_pred = model.fit_predict(X)
```

```python
# Visualize Clusters

plt.scatter(
    X[:,0],
    X[:,1],
    c=y_pred
)

plt.scatter(
    model.cluster_centers_[:,0],
    model.cluster_centers_[:,1],
    s=300,
    c='red'
)

plt.xlabel("Annual Income")

plt.ylabel("Spending Score")

plt.title("Customer Segmentation")

plt.show()
```
